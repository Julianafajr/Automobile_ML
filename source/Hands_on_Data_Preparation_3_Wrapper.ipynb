# Data Preparation III - Feature Selection: **Wrapper Method**

```python
# Read the cleaned dataset (clean_df.csv) into a DataFrame and display the first few rows
import pandas as pd

df_clean = pd.read_csv('clean_df.csv')
display(df_clean.head())
```

#### Convert horsepower-binned from categorical to numeric using Ordinal Encoding


```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(categories=[['Low', 'Medium', 'High']])
encoded_horse = encoder.fit_transform(df_clean[["horsepower-binned"]])

print(f"Encoded horsepower-binned: {encoded_horse}")
```

```python
# merge the new dataframe to the original datafram
encoded_horse_df = pd.DataFrame(encoded_horse, columns=['horsepower-binned_encoded'])
df_clean = pd.concat([df_clean, encoded_horse_df], axis=1)

# drop original column "horsepower-binned" from "df"
df_clean = df_clean.drop('horsepower-binned', axis = 1)
```

```python
df_clean.info()
```

```python
# drop the first column "unnamed" since it is just an ID (not used)
df_clean = df_clean.drop('Unnamed: 0', axis = 1)
```

```python
# Move the target variable "price" to the last column
# Get a list of all columns
cols = df_clean.columns.tolist()

# Remove 'price' from the list
cols.remove('price')

# Append 'price' to the end of the list
cols.append('price')

# Reindex the DataFrame with the new column order
df_clean = df_clean[cols]

# Display the columns to confirm the change
print(df_clean.columns)
```

# Feature Selection Methods
The following are methods for feature selection:

1. Filter: filters features based on the statistical indicators for the feature vectors and those common to the feature vectors and the target variables.
2. Wrapper: attempts to use different feature subsets for modeling and use the model precision as an evaluation indicator for the feature subsets.
3. Embedded: evaluates feature weights during model training and scores the importance of the features.


##2. Wrapper

## Wrapper
In the wrapper selection method, different feature subsets are used for modeling, the model precision
is used as the evaluation indicator for the feature subsets, and a base model is selected to perform multi-round training. After each round of training, features of some weight coefficients are removed, and then the next round of training is performed based on the new feature set. The RFE() method of the feature_selection submodule in sklearn is invoked. The logistic regression model LogisticRegressio() is used as the base model to be invoked, and parameter will be transferred into
this model.

Wrapper:

estimator: basic training model, which is a logistic regression model in this example.

n_features_to_select: indicates the number of retained features.


fit(X,y): invokes and trains a model.

```python
from sklearn.feature_selection import RFE
from sklearn.linear_model import Ridge

# Split the column into X (independent variables/input) and y (target/output)
X = df_clean.drop('price',axis=1)
y = df_clean['price']

x_rfe=RFE(estimator=Ridge(alpha=.5), n_features_to_select=20).fit(X, y)
print(x_rfe.n_features_ )
print(x_rfe.support_ )
print(x_rfe.ranking_ )
print(x_rfe.estimator_ )
```

##Variable Removal
Based on the results of the preceding selection method, the variables that have a small model impact weight are removed.

```python
# Based on the results of the preceding selection method, the variables that have a small model impact weight are removed.
del_cols = X.columns[~x_rfe.support_].tolist()
df_select = df_clean.drop(del_cols,axis=1)
df_select.head()
```
