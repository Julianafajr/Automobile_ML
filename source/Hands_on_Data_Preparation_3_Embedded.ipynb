# Data Preparation III - Feature Selection: **Embedded Method**

```python
# Read the cleaned dataset (clean_df.csv) into a DataFrame and display the first few rows
import pandas as pd

df_clean = pd.read_csv('clean_df.csv')
display(df_clean.head())
```

#### Convert horsepower-binned from categorical to numeric using Ordinal Encoding


```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(categories=[['Low', 'Medium', 'High']])
encoded_horse = encoder.fit_transform(df_clean[["horsepower-binned"]])

print(f"Encoded horsepower-binned: {encoded_horse}")
```

```python
# merge the new dataframe to the original datafram
encoded_horse_df = pd.DataFrame(encoded_horse, columns=['horsepower-binned_encoded'])
df_clean = pd.concat([df_clean, encoded_horse_df], axis=1)

# drop original column "horsepower-binned" from "df"
df_clean = df_clean.drop('horsepower-binned', axis = 1)
```

```python
df_clean.info()
```

```python
# drop the first column "unnamed" since it is just an ID (not used)
df_clean = df_clean.drop('Unnamed: 0', axis = 1)
```

```python
# Move the target variable "price" to the last column
# Get a list of all columns
cols = df_clean.columns.tolist()

# Remove 'price' from the list
cols.remove('price')

# Append 'price' to the end of the list
cols.append('price')

# Reindex the DataFrame with the new column order
df_clean = df_clean[cols]

# Display the columns to confirm the change
print(df_clean.columns)
```

# Embedded
The embedded method uses a machine learning model for training to obtain weight coefficients of
features, and selects features in descending order of the weight coefficients.
Common embedded methods are based on either of the following:

1.  Linear model and regularization
2.  Feature selection of a tree model

In the tree model, the importance of a feature is determined by the depth of the leaf node. In this experiment, the random forest is used to calculate the importance of a feature.

The random forest classification method in the sklearn.ensemble submodule is invoked to train the model by using the fit(X,y) method.

```python
from sklearn.ensemble import RandomForestClassifier

# Split the column into X (independent variables/input) and y (target/output)
X = df_clean.drop('price',axis=1)
y = df_clean['price']

rfc=RandomForestClassifier()
rfc.fit(X,y)
```

```python
cols=[i for i in X.columns]
sorted_feature=sorted(zip(map(lambda x:round(x,4),rfc.feature_importances_),cols),reverse=True)
sorted_feature
```

##Variable Removal
Based on the results of the preceding selection method, the variables that have a small model impact weight are removed.

```python
# Determine a threshold for feature importance (you can adjust this value)
threshold = 0.005

# Get the features to drop based on the threshold
features_to_drop = [feature for importance, feature in sorted_feature if importance < threshold]

# Drop the less important features from the DataFrame
df_select = df_clean.drop(columns=features_to_drop)

# Display the columns of the reduced DataFrame to confirm the changes
print("Original number of columns:", df_clean.shape[1])
print("Number of columns after dropping less important features:", df_select.shape[1])
print("Columns remaining:", df_select.columns.tolist())
```
